# Coarse-Resolved Wave Crossing a Fixed-Mesh-Refinement Interface

This experiment launches the same right-moving, Gaussian-windowed $E_y/B_z$ wave packet inside a single fine patch, then compares refinement ratios 2, 4, and 6 as the packet crosses from the fine mesh onto the coarse periodic mesh.

Success criteria:

- all three runs remain finite through $t=54$;
- every movie contains the initial packet inside the FMR patch and its later transmission through the right interface at $x=80$;
- measured envelope speeds agree with the local discrete Yee prediction;
- all movies use the same physical frame times and color normalization.


```python
from __future__ import annotations

from pathlib import Path
from tempfile import TemporaryDirectory

import jax
import jax.numpy as jnp
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML, Markdown, display
from matplotlib.animation import FFMpegWriter, FuncAnimation, writers
from matplotlib.patches import Rectangle


def find_repo_root():
    current = Path.cwd().resolve()
    for candidate in (current, *current.parents):
        if (candidate / "PyPIC3D").is_dir() and (candidate / "test_notebooks").is_dir():
            return candidate
    raise RuntimeError("Run this notebook from the PyPIC3D FMR_DEV checkout.")


REPO_ROOT = find_repo_root()
OUTPUT_DIR = REPO_ROOT / "test_notebooks"

import sys

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from PyPIC3D.initialization import initialize_simulation
from PyPIC3D.fmr import (
    B_FIELD_LOCATIONS,
    E_FIELD_LOCATIONS,
    prolong_e_to_fine_interface,
)

jax.config.update("jax_enable_x64", True)
mpl.rcParams["animation.embed_limit"] = 100.0

print(f"Repository: {REPO_ROOT}")
print(f"JAX devices: {jax.devices()}")
```

## Experiment design

The root grid is $128\times6\times6$ with unit spacing and periodic boundaries. The fine patch covers coarse indices `[32:80, 1:5, 1:5]`, or the physical box $[32,80]\times[1,5]\times[1,5]$.

The wavelength is 16 coarse cells. Consequently the fine levels resolve it with 32, 64, and 96 cells for refinement ratios 2, 4, and 6. The global timestep is $0.3/r$, holding the fine-grid CFL number fixed at 0.3. One movie frame spans 0.6 time units, so all ratios are compared at exactly the same physical times.

This remains a quasi-one-dimensional plane-wave test. The movie is restricted to the transverse FMR aperture, so it follows fine data inside the patch and root data before and after it without displaying the separately evolving coarse transverse strips.


```python
REFINEMENT_RATIOS = (2, 4, 6)
ROOT_SHAPE = (128, 6, 6)
PATCH_START = (32, 1, 1)
PATCH_STOP = (80, 5, 5)
GUARD_CELLS = 2

WAVE_AMPLITUDE = 1.0
PULSE_CENTER = 56.0
PULSE_WIDTH = 6.0
WAVELENGTH = 16.0
WAVE_NUMBER = 2.0 * np.pi / WAVELENGTH

FINAL_TIME = 54.0
FRAME_DT = 0.6
FRAME_TIMES = np.linspace(0.0, FINAL_TIME, int(round(FINAL_TIME / FRAME_DT)) + 1)
NUM_FRAMES = len(FRAME_TIMES)
PLOT_Z = 3.0

ROOT_X_MIN, ROOT_X_MAX = 0.0, float(ROOT_SHAPE[0])
ROOT_Y_MIN, ROOT_Y_MAX = 0.0, float(ROOT_SHAPE[1])
ROOT_Z_MIN, ROOT_Z_MAX = 0.0, float(ROOT_SHAPE[2])

PATCH_X_MIN, PATCH_Y_MIN, PATCH_Z_MIN = map(float, PATCH_START)
PATCH_X_MAX, PATCH_Y_MAX, PATCH_Z_MAX = map(float, PATCH_STOP)

FINE_SPEED_TIME = (0.0, 9.0)
ROOT_SPEED_TIME = (36.0, 51.0)
SPEED_RELATIVE_TOLERANCE = 0.05

interface_amplitudes = np.abs(
    np.exp(-0.5 * ((np.asarray((PATCH_X_MIN, PATCH_X_MAX)) - PULSE_CENTER) / PULSE_WIDTH) ** 2)
)
assert np.max(interface_amplitudes) < 5.0e-4
assert np.isclose(FRAME_TIMES[-1], FINAL_TIME)
NUM_FRAMES
```

## Production FMR setup and wave initialization

Each ratio is initialized through `initialize_simulation`. The wave is sampled independently at the native staggered $E_y$ and $B_z$ coordinates on both levels. The production prolongation operator then overwrites constrained fine-$E$ interface values before evolution starts.


```python
def fmr_config(refinement_ratio, output_dir):
    dt = 0.3 / refinement_ratio
    number_of_steps = int(round(FINAL_TIME / dt))

    return {
        "static_parameters": {
            "name": f"High-frequency FMR wave, r={refinement_ratio}",
            "output_dir": str(output_dir),
            "solver": "electrodynamic_yee",
            "Nt": number_of_steps,
            "guard_cells": GUARD_CELLS,
            "particle_tile_nx": ROOT_SHAPE[0],
            "particle_tile_ny": ROOT_SHAPE[1],
            "particle_tile_nz": ROOT_SHAPE[2],
            "current_calculation": "j_from_rhov",
            "filter_j": "none",
            "x_bc": "periodic",
            "y_bc": "periodic",
            "z_bc": "periodic",
        },
        "dynamic_parameters": {
            "Nx": ROOT_SHAPE[0],
            "Ny": ROOT_SHAPE[1],
            "Nz": ROOT_SHAPE[2],
            "x_min": ROOT_X_MIN,
            "x_max": ROOT_X_MAX,
            "y_min": ROOT_Y_MIN,
            "y_max": ROOT_Y_MAX,
            "z_min": ROOT_Z_MIN,
            "z_max": ROOT_Z_MAX,
            "dt": dt,
            "C": 1.0,
            "eps": 1.0,
            "mu": 1.0,
        },
        "fmr": {
            "enabled": True,
            "levels": [
                {
                    "parent": 0,
                    "refinement_ratio": refinement_ratio,
                    "coarse_start": list(PATCH_START),
                    "coarse_stop": list(PATCH_STOP),
                }
            ],
        },
    }


def component_axis(grids, locations, axis):
    tiled_grid = grids.tiled_vertex_grid if locations[axis] == "V" else grids.tiled_center_grid
    return tiled_grid[axis][0, 0, 0]


def wave_profile(x):
    displacement = x - PULSE_CENTER
    envelope = jnp.exp(-0.5 * (displacement / PULSE_WIDTH) ** 2)
    return WAVE_AMPLITUDE * envelope * jnp.cos(WAVE_NUMBER * displacement)


def initialize_wave(fields, static_parameters, dynamic_parameters):
    E_levels, B_levels, J_levels = fields[:3]
    initialized_E = []
    initialized_B = []

    for E_level, B_level, level_data in zip(
        E_levels,
        B_levels,
        dynamic_parameters.fmr.levels,
    ):
        x_Ey = component_axis(level_data.grids, E_FIELD_LOCATIONS[1], axis=0)
        x_Bz = component_axis(level_data.grids, B_FIELD_LOCATIONS[2], axis=0)

        Ex, Ey, Ez = E_level
        Bx, By, Bz = B_level
        Ey = jnp.broadcast_to(
            wave_profile(x_Ey)[jnp.newaxis, jnp.newaxis, jnp.newaxis, :, jnp.newaxis, jnp.newaxis],
            Ey.shape,
        )
        Bz = jnp.broadcast_to(
            wave_profile(x_Bz)[jnp.newaxis, jnp.newaxis, jnp.newaxis, :, jnp.newaxis, jnp.newaxis],
            Bz.shape,
        )
        initialized_E.append((Ex, Ey, Ez))
        initialized_B.append((Bx, By, Bz))

    E_levels = tuple(initialized_E)
    B_levels = tuple(initialized_B)
    E_levels = (
        E_levels[0],
        prolong_e_to_fine_interface(
            E_levels[0],
            E_levels[1],
            dynamic_parameters.fmr.levels[1].e_interface_maps,
        ),
    )

    return (E_levels, B_levels, J_levels, *fields[3:])
```

## JIT-compiled evolution and plane sampling

A single compiled scan advances one movie-frame interval at a time. The inner `fori_loop` derives its step count from `FRAME_DT / dt`, giving exactly 0.6 physical time units per frame for every ratio. The active center-$z$ $E_y$ planes are retained for the movies, together with cell-centered $\tfrac12(E_y^2+B_z^2)$ planes for the packet-centroid diagnostic.


```python
def active_component_axis(level, level_data, locations, axis, guard_cells):
    cells = (level.Nx, level.Ny, level.Nz)[axis]
    axis_values = component_axis(level_data.grids, locations, axis)
    return axis_values[guard_cells:guard_cells + cells]


def yee_phase_and_group_velocity(wave_number, spacing, dt, wave_speed=1.0):
    courant = wave_speed * dt / spacing
    half_phase = 0.5 * wave_number * spacing
    omega = 2.0 * np.arcsin(courant * np.sin(half_phase)) / dt
    phase_velocity = omega / wave_number
    group_velocity = (
        wave_speed * np.cos(half_phase)
        / np.sqrt(1.0 - courant**2 * np.sin(half_phase) ** 2)
    )
    return float(phase_velocity), float(group_velocity)


def measure_energy_centroid_speed(times, x, y, energy_history, time_bounds, x_bounds):
    time_mask = (times >= time_bounds[0]) & (times <= time_bounds[1])
    x_mask = (x >= x_bounds[0]) & (x <= x_bounds[1])
    y_mask = (y >= PATCH_Y_MIN) & (y < PATCH_Y_MAX)

    selected_energy = energy_history[:, x_mask, :][:, :, y_mask]
    line_energy = np.mean(selected_energy, axis=2)
    selected_x = x[x_mask]
    energy_sum = np.sum(line_energy, axis=1)
    centroid_numerator = np.sum(line_energy * selected_x[np.newaxis, :], axis=1)
    centroids = np.full_like(energy_sum, np.nan, dtype=float)
    np.divide(centroid_numerator, energy_sum, out=centroids, where=energy_sum > 1.0e-14)

    valid = time_mask & np.isfinite(centroids) & (energy_sum > 1.0e-14)
    speed, intercept = np.polyfit(times[valid], centroids[valid], deg=1)
    fitted_centroids = speed * times + intercept
    return float(speed), centroids, fitted_centroids


def run_ratio(refinement_ratio):
    with TemporaryDirectory(prefix=f"pypic3d_fmr_r{refinement_ratio}_") as output_dir:
        (
            evolve_loop,
            particles,
            fields,
            static_parameters,
            dynamic_parameters,
            _plotting_parameters,
            _plasma_parameters,
            species_config,
        ) = initialize_simulation(fmr_config(refinement_ratio, output_dir))
        assert evolve_loop.__name__ == "time_loop_electrodynamic_fmr_fields"

        fields = initialize_wave(fields, static_parameters, dynamic_parameters)
        steps_per_frame = int(round(FRAME_DT / float(dynamic_parameters.dt)))
        assert np.isclose(steps_per_frame * float(dynamic_parameters.dt), FRAME_DT)

        root_level, fine_level = static_parameters.fmr_levels
        root_data, fine_data = dynamic_parameters.fmr.levels
        g = int(static_parameters.guard_cells)

        root_x = active_component_axis(root_level, root_data, E_FIELD_LOCATIONS[1], 0, g)
        root_y = active_component_axis(root_level, root_data, E_FIELD_LOCATIONS[1], 1, g)
        root_z = active_component_axis(root_level, root_data, E_FIELD_LOCATIONS[1], 2, g)
        fine_x = active_component_axis(fine_level, fine_data, E_FIELD_LOCATIONS[1], 0, g)
        fine_y = active_component_axis(fine_level, fine_data, E_FIELD_LOCATIONS[1], 1, g)
        fine_z = active_component_axis(fine_level, fine_data, E_FIELD_LOCATIONS[1], 2, g)

        root_z_index = int(jnp.argmin(jnp.abs(root_z - PLOT_Z)))
        fine_z_index = int(jnp.argmin(jnp.abs(fine_z - PLOT_Z)))

        def sample_planes(field_state):
            E_levels, B_levels = field_state[:2]
            root_Ey = E_levels[0][1][
                0, 0, 0,
                g:g + root_level.Nx,
                g:g + root_level.Ny,
                g + root_z_index,
            ]
            fine_Ey = E_levels[1][1][
                0, 0, 0,
                g:g + fine_level.Nx,
                g:g + fine_level.Ny,
                g + fine_z_index,
            ]
            root_Bz_left = B_levels[0][2][
                0, 0, 0,
                g:g + root_level.Nx,
                g:g + root_level.Ny,
                g + root_z_index,
            ]
            root_Bz_right = B_levels[0][2][
                0, 0, 0,
                g + 1:g + root_level.Nx + 1,
                g:g + root_level.Ny,
                g + root_z_index,
            ]
            fine_Bz_left = B_levels[1][2][
                0, 0, 0,
                g:g + fine_level.Nx,
                g:g + fine_level.Ny,
                g + fine_z_index,
            ]
            fine_Bz_right = B_levels[1][2][
                0, 0, 0,
                g + 1:g + fine_level.Nx + 1,
                g:g + fine_level.Ny,
                g + fine_z_index,
            ]
            root_energy = 0.5 * (
                root_Ey**2 + 0.5 * (root_Bz_left**2 + root_Bz_right**2)
            )
            fine_energy = 0.5 * (
                fine_Ey**2 + 0.5 * (fine_Bz_left**2 + fine_Bz_right**2)
            )
            return root_Ey, fine_Ey, root_energy, fine_energy

        initial_planes = sample_planes(fields)

        def advance_one_step(field_state):
            _, field_state = evolve_loop(
                particles,
                species_config,
                field_state,
                static_parameters,
                dynamic_parameters,
            )
            return field_state

        def advance_one_frame(field_state, _):
            field_state = jax.lax.fori_loop(
                0,
                steps_per_frame,
                lambda _step, state: advance_one_step(state),
                field_state,
            )
            return field_state, sample_planes(field_state)

        @jax.jit
        def evolve_and_sample(field_state):
            return jax.lax.scan(
                advance_one_frame,
                field_state,
                xs=None,
                length=NUM_FRAMES - 1,
            )

        _final_fields, histories = evolve_and_sample(fields)
        histories = tuple(
            jnp.concatenate((initial[jnp.newaxis], history), axis=0)
            for initial, history in zip(initial_planes, histories)
        )
        root_history, fine_history, root_energy_history, fine_energy_history = histories
        fine_energy_history.block_until_ready()

        result = {
            "ratio": refinement_ratio,
            "dt": float(dynamic_parameters.dt),
            "fine_spacing": float(fine_level.spacing[0]),
            "fine_shape": (fine_level.Nx, fine_level.Ny, fine_level.Nz),
            "points_per_wavelength": int(round(WAVELENGTH / fine_level.spacing[0])),
            "times": FRAME_TIMES.copy(),
            "root_x": np.asarray(jax.device_get(root_x)),
            "root_y": np.asarray(jax.device_get(root_y)),
            "fine_x": np.asarray(jax.device_get(fine_x)),
            "fine_y": np.asarray(jax.device_get(fine_y)),
            "root_Ey": np.asarray(jax.device_get(root_history)),
            "fine_Ey": np.asarray(jax.device_get(fine_history)),
            "root_energy": np.asarray(jax.device_get(root_energy_history)),
            "fine_energy": np.asarray(jax.device_get(fine_energy_history)),
        }
        result["finite"] = bool(
            np.isfinite(result["root_Ey"]).all()
            and np.isfinite(result["fine_Ey"]).all()
            and np.isfinite(result["root_energy"]).all()
            and np.isfinite(result["fine_energy"]).all()
        )
        initial_peak_index = np.unravel_index(
            np.argmax(np.abs(result["fine_Ey"][0])),
            result["fine_Ey"][0].shape,
        )
        initial_peak_x = result["fine_x"][initial_peak_index[0]]
        result["initial_peak_inside_patch"] = bool(
            PATCH_X_MIN <= initial_peak_x < PATCH_X_MAX
        )
        transmitted_x = result["root_x"] >= PATCH_X_MAX + 4.0
        transmitted_y = (result["root_y"] >= PATCH_Y_MIN) & (result["root_y"] < PATCH_Y_MAX)
        result["final_transmitted_peak"] = float(
            np.max(np.abs(result["root_Ey"][-1, transmitted_x, :][:, transmitted_y]))
        )

        root_phase_velocity, root_group_velocity = yee_phase_and_group_velocity(
            WAVE_NUMBER,
            float(dynamic_parameters.dx),
            result["dt"],
        )
        fine_phase_velocity, fine_group_velocity = yee_phase_and_group_velocity(
            WAVE_NUMBER,
            result["fine_spacing"],
            result["dt"],
        )
        fine_measured_velocity, fine_centroids, fine_centroid_fit = measure_energy_centroid_speed(
            result["times"],
            result["fine_x"],
            result["fine_y"],
            result["fine_energy"],
            FINE_SPEED_TIME,
            (PATCH_X_MIN, PATCH_X_MAX),
        )
        root_measured_velocity, root_centroids, root_centroid_fit = measure_energy_centroid_speed(
            result["times"],
            result["root_x"],
            result["root_y"],
            result["root_energy"],
            ROOT_SPEED_TIME,
            (PATCH_X_MAX + 4.0, ROOT_X_MAX - 4.0),
        )
        result.update({
            "root_phase_velocity": root_phase_velocity,
            "root_group_velocity": root_group_velocity,
            "root_measured_velocity": root_measured_velocity,
            "root_centroids": root_centroids,
            "root_centroid_fit": root_centroid_fit,
            "fine_phase_velocity": fine_phase_velocity,
            "fine_group_velocity": fine_group_velocity,
            "fine_measured_velocity": fine_measured_velocity,
            "fine_centroids": fine_centroids,
            "fine_centroid_fit": fine_centroid_fit,
        })

    return result
```

## Run the three refinement ratios

This cell performs three separate JAX compilations because each fine level has a different static shape. Compilation can dominate the first run for each ratio.


```python
results = {}
for ratio in REFINEMENT_RATIOS:
    print(f"Running refinement ratio {ratio}...")
    results[ratio] = run_ratio(ratio)

assert all(result["finite"] for result in results.values())
assert all(result["initial_peak_inside_patch"] for result in results.values())
assert all(result["final_transmitted_peak"] > 1.0e-8 for result in results.values())
assert all(np.array_equal(result["times"], FRAME_TIMES) for result in results.values())
assert FRAME_TIMES[0] == 0.0 and FRAME_TIMES[-1] > PATCH_X_MAX - PULSE_CENTER

for result in results.values():
    result["fine_speed_agrees"] = bool(np.isclose(
        result["fine_measured_velocity"],
        result["fine_group_velocity"],
        rtol=SPEED_RELATIVE_TOLERANCE,
        atol=0.0,
    ))
    result["root_speed_agrees"] = bool(np.isclose(
        result["root_measured_velocity"],
        result["root_group_velocity"],
        rtol=SPEED_RELATIVE_TOLERANCE,
        atol=0.0,
    ))

visible_field_limits = []
for result in results.values():
    root_y_visible = (result["root_y"] >= PATCH_Y_MIN) & (result["root_y"] < PATCH_Y_MAX)
    fine_y_visible = (result["fine_y"] >= PATCH_Y_MIN) & (result["fine_y"] < PATCH_Y_MAX)
    visible_field_limits.extend((
        float(np.max(np.abs(result["root_Ey"][:, :, root_y_visible]))),
        float(np.max(np.abs(result["fine_Ey"][:, :, fine_y_visible]))),
    ))
COMMON_FIELD_LIMIT = max(visible_field_limits)

configuration_table = [
    "| Ratio | dt | Fine shape | Fine dx | Cells / wavelength | Frames | Finite |",
    "|---:|---:|:---:|---:|---:|---:|:---:|",
]
for ratio in REFINEMENT_RATIOS:
    result = results[ratio]
    configuration_table.append(
        f"| {ratio} | {result['dt']:.5f} | {result['fine_shape']} | "
        f"{result['fine_spacing']:.5f} | {result['points_per_wavelength']} | "
        f"{len(result['times'])} | {result['finite']} |"
    )
display(Markdown("\n".join(configuration_table)))

velocity_table = [
    "| Ratio | Region | Yee phase speed | Yee group speed | Measured centroid speed | Within 5% |",
    "|---:|:---|---:|---:|---:|:---:|",
]
for ratio in REFINEMENT_RATIOS:
    result = results[ratio]
    velocity_table.extend((
        f"| {ratio} | fine | {result['fine_phase_velocity']:.5f} | "
        f"{result['fine_group_velocity']:.5f} | {result['fine_measured_velocity']:.5f} | "
        f"{result['fine_speed_agrees']} |",
        f"| {ratio} | root | {result['root_phase_velocity']:.5f} | "
        f"{result['root_group_velocity']:.5f} | {result['root_measured_velocity']:.5f} | "
        f"{result['root_speed_agrees']} |",
    ))
display(Markdown("\n".join(velocity_table)))
print(f"Shared movie color range: [{-COMMON_FIELD_LIMIT:.6g}, {COMMON_FIELD_LIMIT:.6g}]")
```

## Composite FMR movie helper

The root field is masked wherever the fine patch covers the plotted plane. The fine field is drawn over that hole at its native resolution. The viewport is restricted to the patch's transverse aperture, removing the unrelated coarse strips above and below it. All three movies share the same axes and color scale.


```python
def make_composite_movie(result, field_limit):
    root_x = result["root_x"]
    root_y = result["root_y"]
    fine_x = result["fine_x"]
    fine_y = result["fine_y"]

    root_inside_patch = (
        (root_x[:, np.newaxis] >= PATCH_X_MIN)
        & (root_x[:, np.newaxis] < PATCH_X_MAX)
        & (root_y[np.newaxis, :] >= PATCH_Y_MIN)
        & (root_y[np.newaxis, :] < PATCH_Y_MAX)
    )

    def masked_root_frame(frame_index):
        return np.ma.masked_where(root_inside_patch, result["root_Ey"][frame_index]).T

    fig, ax = plt.subplots(figsize=(12, 4), constrained_layout=True)
    coarse_mesh = ax.pcolormesh(
        root_x,
        root_y,
        masked_root_frame(0),
        shading="nearest",
        cmap="RdBu_r",
        vmin=-field_limit,
        vmax=field_limit,
    )
    fine_mesh = ax.pcolormesh(
        fine_x,
        fine_y,
        result["fine_Ey"][0].T,
        shading="nearest",
        cmap="RdBu_r",
        vmin=-field_limit,
        vmax=field_limit,
    )
    patch_outline = Rectangle(
        (PATCH_X_MIN, PATCH_Y_MIN),
        PATCH_X_MAX - PATCH_X_MIN,
        PATCH_Y_MAX - PATCH_Y_MIN,
        fill=False,
        edgecolor="black",
        linewidth=1.5,
        linestyle="--",
        label="FMR patch",
    )
    ax.add_patch(patch_outline)

    title = ax.set_title("")
    ax.set_xlim(ROOT_X_MIN, ROOT_X_MAX)
    ax.set_ylim(PATCH_Y_MIN, PATCH_Y_MAX)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.legend(loc="upper right")
    colorbar = fig.colorbar(coarse_mesh, ax=ax, pad=0.02)
    colorbar.set_label(r"$E_y$")

    def update(frame_index):
        coarse_mesh.set_array(masked_root_frame(frame_index).ravel())
        fine_mesh.set_array(result["fine_Ey"][frame_index].T.ravel())
        title.set_text(
            f"Coarse-resolved wave crossing: r={result['ratio']}, "
            f"fine dx={result['fine_spacing']:.3f}, t={result['times'][frame_index]:.1f}"
        )
        return coarse_mesh, fine_mesh, title

    movie = FuncAnimation(
        fig,
        update,
        frames=len(result["times"]),
        interval=50,
        blit=False,
    )
    return fig, movie


def render_movie(result, field_limit):
    if not writers.is_available("ffmpeg"):
        raise RuntimeError("ffmpeg is required to save the requested MP4 movies.")

    fig, movie = make_composite_movie(result, field_limit)
    movie_path = OUTPUT_DIR / f"fmr_wave_ratio_{result['ratio']}.mp4"
    movie.save(movie_path, writer=FFMpegWriter(fps=20, bitrate=2400), dpi=110)
    inline_movie = HTML(movie.to_jshtml(default_mode="loop"))
    plt.close(fig)

    print(f"Saved {movie_path} ({movie_path.stat().st_size / 1.0e6:.2f} MB)")
    display(inline_movie)
    return movie_path
```

## Numerical acceptance result

The target is agreement between the energy-centroid speeds and the local Yee group velocities to within 5%. This tolerance covers finite packet bandwidth, centroid fitting, and interface transients without replacing the discrete-dispersion prediction by an exact-$c$ assumption. A failed result is reported explicitly because it is evidence about the production FMR operator, not a reason to hide the movies or prevent this diagnostic notebook from finishing.


```python
all_speeds_agree = all(
    result["fine_speed_agrees"] and result["root_speed_agrees"]
    for result in results.values()
)
if all_speeds_agree:
    print("PASS: all measured envelope speeds agree with the local Yee predictions.")
else:
    display(Markdown(
        "**FAIL:** at least one measured envelope speed differs from its local Yee "
        "group-velocity prediction by more than 5%. The table above is the "
        "numerical result; the production FMR source is intentionally unchanged here."
    ))
```

## Interpretation checklist

Compare the movies at the same displayed time, especially around $t\approx24$ when the packet center reaches the right interface. Look for:

- transmitted phase and amplitude on the coarse mesh;
- left-moving reflections remaining in the fine patch;
- interface-localized distortion or transverse structure;
